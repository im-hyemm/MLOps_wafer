```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
wafer_df = pd.read_csv('data/wafer_df.csv')
wafer_df.head()
```

```python
failure_type_counts = wafer_df['failureType'].value_counts()

# Calculate percentages for the legend
percentages = (failure_type_counts / failure_type_counts.sum()) * 100
legend_labels = [f'{label} ({p:.1f}%)' for label, p in zip(failure_type_counts.index, percentages)]

plt.figure(figsize=(12, 10)) # Adjust figure size to accommodate legend
plt.pie(failure_type_counts, startangle=90, colors=sns.color_palette('pastel'))
plt.title('Distribution of Failure Types')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.

# Add legend outside the pie chart
plt.legend(legend_labels, title="Failure Types", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
plt.show()
```
